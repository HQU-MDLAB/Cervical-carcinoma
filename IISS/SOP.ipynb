```python
import matplotlib.pyplot as plt
import pandas as pd 
import seaborn as sns
from PIL import Image
import re
import os
os.chdir("G:/data/RENNAO/20250508-result")
os.getcwd()
```

```python
### 定义函数
def SOP(dapi_file, transcripts_file):
    
    ### DAPI图
    dapi = Image.open(dapi_file)
    
    ### 信号点
    transcripts = pd.read_csv(transcripts_file, index_col = 0)
    
    ### 自定义基因颜色
    gene_colors = {
        'PLA': 'red',
        'RBFOX': 'blue',
        'MBP': 'green',
        'DapB': 'grey',
        'GFAP': 'cyan',
        'TYROBP': 'purple',
        '0': 'black'
    }
    
    ### 画图
    plt.figure(figsize=(8, 8))
    plt.imshow(dapi, cmap='gray')
    plt.axis('off')
    sns.scatterplot(
        data=transcripts,
        x='x',
        y='y',
        hue='gene',
        palette=gene_colors,
        # palette='tab20',
        edgecolor=None,
        s=10,
        alpha=0.8,
        legend='auto'
    )
    plt.legend(loc='lower right', frameon=True, fontsize=10, markerscale=1.2)

    sample_name = re.split(r'\\', dapi_file, maxsplit=6)[1]
    output_dir = '20250513_CJY_outs'
    os.makedirs(output_dir, exist_ok=True)
    output_path = os.path.join(output_dir, sample_name + '.png')
    plt.savefig(output_path, dpi=300, transparent=False, bbox_inches='tight')
    plt.close() 

### 列出目录中所有 CellSegBackground.png 和 transcripts.csv 的完整路径
def list_files(directory):

    cellseg_files = []
    transcripts_files = []
    
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file == 'CellSegBackground.png':
                cellseg_files.append(os.path.join(root, file))
            elif file == 'transcripts.csv':
                transcripts_files.append(os.path.join(root, file))
    
    return cellseg_files, transcripts_files
cellseg_files, transcripts_files = list_files('G:/data/RENNAO/20250508-result')
assert len(cellseg_files) == len(transcripts_files), "文件数量不一致！"

### 批量运行
for cellseg_file, transcripts_file in zip(cellseg_files, transcripts_files):
    SOP(cellseg_file, transcripts_file)
    
```
